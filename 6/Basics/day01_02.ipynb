```python
def calculate_average(*numbers):

    total = len(numbers)
    total_sum = sum(numbers)
    return total_sum / total


def stats(*nums):
    maximum = max(nums)
    minimum = min(nums)
    avg = calculate_average(*nums)

    return maximum, minimum, avg


```
