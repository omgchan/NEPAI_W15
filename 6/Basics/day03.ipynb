## 2

```python
student = {
    "name": "Chandan",
    "math": 85,
    "python": 92,
    "statistics": 78
}

total_sum =0
total_count =0

for key, value in student.items():

    if isinstance(value, (int, float)):
        total_sum += value
        total_count += 1

    

average = total_sum / total_count
print(f"Average score: {average}")
    

```
