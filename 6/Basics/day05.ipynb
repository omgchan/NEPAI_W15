## 2

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age 
    def intro(self):
        return f"Introduction"

class Student(Person):
    def __init__(self, name, age, major):
        super().__init__(name, age)
        self.major = major


    def intro(self):
        return f"Name: {self.name} \nAge: {self.age} \nMajor: {self.major}" 

s = Student("chandan", 20, "CSIT")
print(s.intro())

s2 = Student("vomjan", 30, "AGE")
print(s2.intro())

```

```python
with open("marks.txt", 'w') as file:
    file.write("85\n")
    file.write("72\n")
    file.write("91\n")
    file.write("64\n")
    file.write("88\n")

class FileStats:
    def __init__(self, filename):
        self.filename = filename
        self.marks = []

    def open(self):
        with open(self.filename, 'r') as file:
            for line in file:
                self.marks.append(int(line.strip()))

    def average(self):
        self.open()
        return sum(self.marks) / len(self.marks)

s = FileStats("marks.txt")
s.average()


```

```python
class FileStats:
    def __init__(self, filename):
        self.filename = filename
        self.marks = []

    def open(self):
        try: 
            with open(self.filename, 'r') as file:
                for line in file:
                    self.marks.append(int(line.strip()))

        except FileNotFoundError as e :
            return False

        return True

    def average(self):
        if not self.open():
            return f"Sorry File Not Found: {self.filename}"
        if not self.marks:
            return "No marks available to calculate average."
        return sum(self.marks) / len(self.marks)
            


s = FileStats("a.txt")
s.average()
```
