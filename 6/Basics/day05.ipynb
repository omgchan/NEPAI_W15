## 1

```python
class Animal:
    def speak(self):
        return "Animal Noises"


class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

dog = Dog()
print(dog.speak())
```

## 2

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age 
    def intro(self):
        return f"Introduction"

class Student(Person):
    def __init__(self, name, age, major):
        super().__init__(name, age)
        self.major = major


    def intro(self):
        return f"Name: {self.name} \nAge: {self.age} \nMajor: {self.major}" 

s = Student("chandan", 20, "CSIT")
print(s.intro())

s2 = Student("vomjan", 30, "AGE")
print(s2.intro())

```
